# Preprocesamiento de datos

> **Nota:** dataset ficticio de 8 clientes, creado para practicar la técnica. Sirve para ilustrar el procedimiento, no para sacar conclusiones generalizables.

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler, StandardScaler, LabelEncoder
```

## 1. Cargar datos y simular valores faltantes

```python
# Crear DataFrame base
df = pd.DataFrame({
    "Cliente": [1, 2, 3, 4, 5, 6, 7, 8],
    "Edad": [25, 40, 35, 50, 30, 45, 38, 29],
    "Ingresos": [50000, 60000, 70000, 80000, 45000, 85000, 72000, 48000],
    "Historial_Pagos": ["Bueno", "Regular", "Malo", "Bueno", "Regular", "Malo", "Bueno", "Regular"],
    "Categoría": ["A", "B", "C", "A", "B", "C", "A", "B"]
})

# Simular valores nulos
df.loc[2, 'Edad'] = np.nan
df.loc[7, 'Ingresos'] = np.nan

df
```

## 2. Imputación de valores faltantes

```python
# Imputar con mediana y media
df["Edad"] = df["Edad"].fillna(df["Edad"].median())
df["Ingresos"] = df["Ingresos"].fillna(df["Ingresos"].mean())
df
```

## 3. Detección y eliminación de outliers en Ingresos (IQR)

```python
Q1 = df["Ingresos"].quantile(0.25)
Q3 = df["Ingresos"].quantile(0.75)
IQR = Q3 - Q1
limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

df = df[(df["Ingresos"] >= limite_inferior) & (df["Ingresos"] <= limite_superior)]
df
```

## 4. Transformación de datos numéricos

```python
scaler_minmax = MinMaxScaler()
df["Ingresos_Normalizado"] = scaler_minmax.fit_transform(df[["Ingresos"]])

scaler_standard = StandardScaler()
df["Edad_Estandarizada"] = scaler_standard.fit_transform(df[["Edad"]])
df
```

## 5. Codificación de variables categóricas

```python
df = pd.get_dummies(df, columns=["Historial_Pagos"], prefix="Historial")
le = LabelEncoder()
df["Categoría_Label"] = le.fit_transform(df["Categoría"])
df
```
