# Análisis - Visualización de Datos
Actividad integradora sobre dataset de clientes de una tienda en línea.

> **Nota:** dataset ficticio de 8 clientes de una tienda en línea, creado para practicar la técnica. Sirve para ilustrar el procedimiento, no para sacar conclusiones generalizables.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Crear DataFrame
df = pd.DataFrame({
    "Cliente": [1, 2, 3, 4, 5, 6, 7, 8],
    "Edad": [25, 40, 35, 50, 30, 45, 38, 29],
    "Ingresos": [50000, 60000, 70000, 80000, 45000, 85000, 72000, 48000],
    "Compras": [5, 15, 10, 20, 7, 18, 12, 6],
    "Categoría": ["A", "B", "C", "A", "B", "C", "A", "B"]
})
df
```

## 1. Análisis Exploratorio de Datos (EDA)
Se presentan histogramas, boxplot y scatter plot para analizar la distribución y relaciones de los datos.

```python
# Histograma de Edades
sns.histplot(df["Edad"], bins=5, kde=True)
plt.title("Distribución de Edades")
plt.show()

# Histograma de Ingresos
sns.histplot(df["Ingresos"], bins=5, kde=True)
plt.title("Distribución de Ingresos")
plt.show()

# Boxplot de Compras
sns.boxplot(x=df["Compras"])
plt.title("Boxplot de Compras")
plt.show()

# Scatter Plot Ingresos vs Compras
sns.scatterplot(x=df["Ingresos"], y=df["Compras"], hue=df["Categoría"])
plt.title("Relación entre Ingresos y Compras")
plt.show()
```

## 2. Visualización Avanzada
Aplicamos técnicas como Pairplot, Heatmap y Violin Plot.

```python
# Pairplot
sns.pairplot(df[["Edad", "Ingresos", "Compras"]])
plt.show()

# Heatmap de Correlaciones
correlaciones = df.corr(numeric_only=True)
sns.heatmap(correlaciones, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Mapa de Correlación de Variables")
plt.show()

# Violin Plot de Compras por Categoría
sns.violinplot(x="Categoría", y="Compras", data=df, palette="muted")
plt.title("Distribución de Compras por Categoría")
plt.show()
```

## 3. Conclusiones
- **Distribución:** las edades e ingresos están relativamente dispersos.
- **Relación:** mayores ingresos tienden a asociarse con mayor cantidad de compras.
- **Correlaciones:** son fuertes: ingresos–compras r ≈ 0.86 y edad–compras r ≈ 0.99 (con solo 8 registros, estos valores son muy inestables).
- **Distribuciones:** la categoría A muestra mayor variabilidad en compras.
